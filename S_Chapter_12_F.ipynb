<h3>1. Encapsulation</h3>
<pre>
    <p>Encapsulation is a mechanism where the data (variables) and the code (methods) that act
on the data will bind together. For example, if we take a class, we write the variables and
methods inside the class. Thus, class is binding them together.</p>
</pre>
<img src="Enc.png">

```python
# a class is an example for encapsulation
class Student:
    # to declare and initialize the variables
    def __init__(self):
        self.id = 10
        self.name = 'Raju'
        
        # display students details
        def display(self):
            print(self.id)
            print(self.name)
```

<h3>2. Abstraction </h3>
<pre><p>There may be a lot of data, a class contains and the user does not need the entire data.
The user requires only some part of the available data.</p></pre>
<img src="oop3.png">

```python
class Myclass:
    # this is constructor.
    def __init__(self):
        self.x = 5  # public variable
        self.__y = 7 # this is private variable
    # instance method to access variables
    def display(self):
        print(self.x)  # x is vailable directly
        print(self._Myclass__y)

print('Accessing variable through method:')
o1 = Myclass()
o1.display()
print('Accessing variables through instance:')
print(o1.x)
print(o1._Myclass__y)
        

print(o1.y) # error
```

<img src="oops4.png">

#  Inheritance

<img src="SIMI.jpg">




```python
# Creating class B from class A: Single inheritance
class Vehicle :
    a = 1
    b = 2
    def method1(cls):
        print(cls.a)
        print(cls.b)
class Car(Vehicle):
    c = 3
    def method2(cls):
        print(cls.c)
        


s = Car()
print(s.a)
print(s.b)
print(s.c)
```

```python
# Creating class C from class A and Class B:Multiple inheritance
class Car :
    a = 1
    b = 2
    def method1(cls):
        print(cls.a)
        print(cls.b)
class Bus :
    c = 3
    def method2(cls):
        print(cls.c)
        
class Vehicle(Car,Bus):
    pass

s = Vehicle()
print(s.a)
print(s.b)
print(s.c)
```
