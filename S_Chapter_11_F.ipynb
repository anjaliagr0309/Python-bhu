```python
# creating a dictionary from the keyboard
x = {} # take an empty dictionary
print('How many elements? ', end='')
n = int(input()) # n indicates no. of key-value pairs
for i in range(n): # repeat for n times
    print('Enter key: ', end='')
    k = input() # key is string
    print('Enter its value: ', end='')
    v = int(input()) # value is integer
    x.update({k:v}) # store the key-value pair in dictionary x

# display the dictionary
print('The dictionary is: ', x)

# find sum of vlaues 
sum = 0
for value in x.values():
    sum = sum + value
print('Sum of values in the dictionary: ', sum) # display s



# display only keys
print('Keys in dict= ', x.keys())
# display only values
print('Values in dict= ', x.values())
# display both key and value pairs as tuples
print('Items in dict= ', x.items())



# accept a name from keyboard
print('Enter key: ', end='')
key = input()

# find the value of respective key
# retrieve value by giving the key
value = x.get(key, -1)
if(value == -1):
    print('name not found')
else:
    print('{}  and its value {}.'.format(key, value))
    
    
    
# want to retrive both key and value   
for k, v in x.items():
    print('Key= {} Value= {}'. format(k, v))
```

```python
# Finding how many times each letter is repeated in a string.
# take a string with some letters
str = "Book"

# take an empty dictionary
dict1 = {}

# store into dict each letter as key and its
# number of occurrences as value
for x in str:
    dict1[x] = dict1.get(x, 0) + 1
   
# Display key and value pairs of dict

for k, v in dict1.items():
    print('Key = {}\t Its occurrences= {}'.format(k, v))
```

```python
# converting lists into a dictionary
# take two separate lists with elements
countries = ["USA", "India", "Germany", "France"]
cities = ['Washington', 'New Delhi', 'Berlin', 'Paris']

# make a dictionary
z = zip(countries, cities)
d = dict(z)

# display key - value pairs from dictionary d
print('{:15s} -- {:15s}'.format('COUNTRY', 'CAPITAL'))
for k in d:
    print('{:15s} -- {:15s}'.format(k, d[k]))
```

```python
# converting a string into a dictionary
# take a string
str = "I=23,II=20,III=19,IV=22"
# brake the string at ',' and then at '='
# store the pieces into a list lst

lst=[]
for x in str.split(','):
    y= x.split('=')
    lst.append(y)

    
print(lst)
    
# convert the list into dictionary 'd'
# but this 'd' will have both name and age as strings

d = dict(lst)

# create a new dictionary 'd1' with name as string
# and age as integer

d1={}
for k, v in d.items():
    d1[k] = int(v)
# display the final dictionary
print(d1)
```

```python
# {key_expression: value_expression for item in iterable}
# dictionary comprehension 
s = "I=23,II=20,III=19,IV=22"

d = {name: int(age) for name, age in (x.split('=') for x in s.split(','))}
print(d)


# Split the string by commas

#Split each item by '='

# Unpack into name and age
```

* set does not support indexing 
* Sets support many mathematical fucntions, such as set union and set intersection

* Mathematical Operations of Sets: union,intersection,diffrence,equivalnce

```python
cities = city_UP.union(city_MP_UP)   
cities
```

```python
type(cities)
```

```python
cities2 = city_UP.intersection(city_MP_UP)  # & operator 
```

```python
cities2
```

```python
#diffrence
cities3 = city_UP - city_MP_UP   # diffrence operator
cities3
```

```python
# symmetric diffrence

cities5 = city_UP ^ city_MP_UP
```

```python
cities5
```
