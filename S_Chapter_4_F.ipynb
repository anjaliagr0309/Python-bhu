<h3>1. Arithmetic Operators </h3>

<img src="IMages/Arithmetic Operators.png">

```python
x,y = 19,3
print("x =", x)
print("y =", y)
print("x + y = ",x+y)  # Addition
print("x - y =",x-y)  # Subtraction
print("x / y =",x/y)  # Division
print("x * y =",x*y)  # Multiplication
print("x % y =",x%y)  # Modulo Division
print("x ** y =",x**y) # Exponetiation
print("x // y =",x//y) # Floor Division
```

<h3> Identity Operators:</h3>

used to compare the objects, not if they are equal, but if they are actually the same object, with the same memory location

<img src="is.png">

```python
a = 100
b = 20
id(a)
```

```python
id(20)
```

```python
a=25
b=25
if(a is b):
    print("a and b have same identity")
else:
    print("a and b do not have same identity")
```

```python
x = [10,20,30,40]
y = [10,20,30,40]
if(x is y):
    print("x and y are same")
else:
    print("x and y are not same")
```

```python
id(x)
```

```python
id(y)
```

```python
if (x == y):
    print("both list are same")
else:
    print("both list are diffrent")
```
