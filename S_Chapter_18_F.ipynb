| ---------- | -----------------------------  |

| `[a-z]`    | Characters a to z              |
| `[0-9]`    | Numbers 0 to 9                 |
| `\w`       | Any Alphanumeric character     |

| `(…)`      | Capture Group                  |

**compile(pattern, flags=0)
            - creates a pattern object from regular expression**
        

```python

# to write RE, we should import re: compile(),search(), match(), findall(), split()

import re 

REX = re.compile(r'm\w\w') # may say RE are strings
str = 'mat bat cat rat hat tom'
result = REX.search(str)
print(result.group()) # Returns the matched text.
print(result.start())# Returns the starting index.
print(result.end()) # Returns the ending index (exclusive).
print(result.span()) # Returns (start, end) as a tuple.
```

 # search() / match() / findall()

**findall(pattern, string)
            - Returns all occurrences of a pattern as a list**

```python
#a regular expression to search for strings starting with m and having total 3 characters using the findall() method. 
import re
str = 'man sun mop run fun bun hat tom mat pen ten the not' 
result = re.findall(r'm\w\w', str)
print(result)


for s in result:
    print(s)
```

```python
# a regular expression to retrieve all words starting with a in a given string. 
import re
str = 'Ananya ate a banana at a park'
result = re.findall(r'\ba[\w]*\b', str)
#result = re.findall(r'a[\w]*', str)
# findall() returns a list, retrieve the elements from list
for word in result:
    print(word)
```

**These are called quantifiers in regular expressions. They specify how many times a character or pattern should occur**

| Pattern  | Meaning              |
| -------- | -------------------- |
| `a*`     | 0 or more `a`s       |
| `a+`     | 1 or more `a`s       |
| `a?`     | 0 or 1 `a`           |
| `a{3}`   | Exactly 3 `a`s       |
| `a{2,}`  | 2 or more `a`s       |
| `a{2,5}` | Between 2 and 5 `a`s |


```python
# a regular expression to retrieve all words having 2 or more a's
import re

text = "Aarav has aaaa apples and aa bananas haa."
print(re.findall(r'a{2,}', text))

# \b represents a word boundary.
```

```python

# retrieve all words having one or more consecutive a's
import re

print(re.findall(r'\ba+\b', text))
```

```python
#between 2 and 5 consecutive a's
print(re.findall(r'\ba{2,5}\b', text))
```

```python
# exactly 3 consecutive a's
print(re.findall(r'a{3}', text))
```

```python
# zero or more a's
print(re.findall(r'a*', text))

# Because * allows zero occurrences, it matches at almost every position in the string, producing many empty strings.
```

```python
# zero or one a
print(re.findall(r'a?', text))

# Because ? matches either one a or nothing.
```

```python
import re

text = "Aarav has aaaa apples and aa bananas"

print(re.findall(r'a+', text))      # ['aa', 'a', 'aaaa', 'a', 'aa', 'a']
print(re.findall(r'a{2,}', text))   # [ 'aaaa', 'aa']
print(re.findall(r'a{2,5}', text))  # [ 'aaaa', 'aa']
print(re.findall(r'a{3}', text))    # ['aaa']
```

```python
#  a regular expression that reads email-ids from a text file
import re
# open the file for reading
f = open('mails.txt', 'r')

# repeat for each line of the file
for line in f:
    res = re.findall(r'\S+@\S+', line)  # S+ represent several characters, id have some char before n after @
    res1 = re.search(r'\d+', line)
    
    # display if there are some elements in result
    if len(res)>0:
        print(res)
print(res1.group())
    
# close the file
f.close()
```

**split(pattern, string, maxsplit=0)
            - Splits a string wherever the pattern occurs**

```python
#  a regular expression to split a string into pieces where one or more non alpha numeric characters are found. ¶

import re
str = 'This; is the: "Core" book of Python Essentials: learn, build and apply'
result = re.split(r'\W+', str)  # w for A-Z,a-z,0-9   
print(result)
```

**sub(pattern, repl, string, count=0)
            - Replaces occurrences of a pattern with another string**

```python
# a regular expression to replace a string with a new string.
import re
str = 'Python Essentials: learn, build and teach '
res = re.sub(r'teach', 'apply', str)
print(res)
```

```python

# a regular expression to search whether a given string is starting with ‘He’ or not. ¶
import re
str = "Hello World"
res = re.search(r"^He", str)# the carat symbol is used to check abt substring starting with
if res:
    print ("String starts with 'He'")
else:
    print("String does not start with 'He'")
```

```python

#  a regular expression to search for a word at the ending of a string.
import re
str = "Hello World"
res = re.search(r"World$", str)# substring ending with $
if res:
    print ("String ends with 'World'")
else:
    print("String does not end with 'World'")
```
