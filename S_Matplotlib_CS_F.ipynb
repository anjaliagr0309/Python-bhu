```python
#https://matplotlib.org/
# Line Plot
import numpy as np
from matplotlib import pyplot as plt
```

```python
x = np.arange(1,11) # numpy array
x
```

```python
y = x*2
y
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y)
plt.title("First line plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
plt.plot(x,y, color='orange', linestyle=':', linewidth=4)
plt.show()
```

```python
#adding two lines in same plot

x = np.arange(1,11)
y1= 2*x
y2 = 3*x
```

```python
plt.plot(x,y1,color='g',linestyle=':',linewidth=2)
plt.plot(x,y2,color='r',linestyle='-',linewidth=3)
plt.title("line Plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid(True)
plt.show()
```

```python
# adding sub plots

x = np.arange(1,11)
y1= 2*x
y2 = 3*x

plt.subplot(1,2,1)
plt.plot(x,y1,color='g',linestyle=':',linewidth=2)

plt.subplot(1,2,2)
plt.plot(x,y2,color='r',linestyle='-',linewidth=3)

plt.show()
```

```python
plt.subplot(2,1,1)
plt.plot(x,y1,color='g',linestyle=':',linewidth=4)

plt.subplot(2,1,2)
plt.plot(x,y2,color='r',linestyle='-',linewidth=1)

plt.show()
```

```python
# barplot  # frequency chart for the qualitative data, help us to understand the distribution of categorical col

student = {"Bob":87,"Matt":56,"sam":27}
```

```python
names = list(student.keys())
values = list(student.values())
plt.bar(names,values)
plt.show()
```

```python
# adding titles and labels
names = list(student.keys())
values = list(student.values())
plt.bar(names,values,color = 'r')
plt.title("Bar Plot")
plt.xlabel("Names")
plt.ylabel("Marks")
plt.grid(True)

plt.show()
```

```python
# horizantal bar plot
plt.barh(names,values,color = 'g')
plt.title(" Horizantal Bar Plot")
plt.xlabel("Marks")
plt.ylabel("Names")
plt.grid(True)

plt.show()
```

```python
# scatter plot
x=[10,20,30,40,50,60,70,80,90]
a = [8,1,7,2,0,3,7,3,2]
# both list should have same number of values, it is 9 here
plt.scatter(x,a)
plt.show()
```

```python
x=[10,20,30,40,50,60,70,80,90]
a = [8,1,7,2,0,3,7,3,2]
plt.scatter(x,a, marker="*",c="r", s=500)
plt.show()
```

```python
# adding two marker in same plot
x=[10,20,30,40,50,60,70,80,90]
a = [8,1,7,2,0,3,7,3,2]
b= [7,4,6,1,8,5,3,2,0]
plt.scatter(x,a, marker="*",c="r", s=100)
plt.scatter(x,b, marker=".",c="g", s=200)
plt.show()
```

```python
# adding subplot
x=[10,20,30,40,50,60,70,80,90]
a = [8,1,7,2,0,3,7,3,2]
b= [7,4,6,1,8,5,3,2,0]
plt.subplot(1,2,1)
plt.scatter(x,a, marker="*",c="r", s=100)
plt.subplot(1,2,2)
plt.scatter(x,b, marker=".",c="g", s=200)
plt.show()
```

```python
# adding subplot
x=[10,20,30,40,50,60,70,80,90]
a = [8,1,7,2,0,3,7,3,2]
b= [7,4,6,1,8,5,3,2,0]
plt.subplot(2,1,1)
plt.scatter(x,a, marker="*",c="r", s=100)
plt.subplot(2,1,2)
plt.scatter(x,b, marker=".",c="g", s=200)
plt.show()
```

```python
# histogram  # frequency distribution of a set of continuous numerical data/variable
data = [1,2,3,3,3,3,9,9,5,4,4,4,8,8,8,8,7]
plt.hist(data)  # invoke hist() method from matplotlib
plt.show()
```

```python
# changing Aesthetics
data = [1,2,3,3,3,3,9,9,5,4,4,4,8,8,8,8,7]
plt.hist(data, color='g',bins= 4)
plt.show()
```

```python
# Box Plot, gives 5 point summary, min,25,50,75 and 100%

one = [1,2,3,4,5,6,7,8,9]
two = [1,2,3,4,5,4,3,2,1]
three = [6,7,8,9,8,7,6,5,4]

data = list([one,two,three])


plt.boxplot(data)
plt.show()
```

```python
# Pie Chart, help us to understand the frequncy or percentage of the categorical value

Fruit = ['Apple','Oranges','Mango','Guava']
Quantity = [67,34,100,29]
plt.pie(Quantity,labels = Fruit)
plt.show()
```

```python
Fruit = ['Apple','Oranges','Mango','Guava']
Quantity = [67,34,100,29]
plt.pie(Quantity,labels = Fruit, autopct = '%0.1f%%',colors = ['yellow','grey','blue','pink'])
plt.show()
```

```python
# DoughNut Chart
Fruit = ['Apple','Oranges','Mango','Guava']
Quantity = [67,34,100,29]
plt.pie(Quantity,labels = Fruit, radius=2)
plt.pie([1],colors=['w'],radius=1)
plt.show()
```
