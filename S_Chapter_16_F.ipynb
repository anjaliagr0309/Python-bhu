An exception is an error that occurs during the execution of a program. When an exception occurs, Python stops the normal flow of the program and raises an error.

**Error in Python can be of three types**
* ***Syntax errors*** are the problems in a program due to which the program will stop the execution.
* ***Runtime errors (Exception)*** PVM cannot execute the byte code
* ***Logical Errors ***  are raised when some internal events occur which changes the normal flow of the program. 

**Some of the common Exception Errors are :** 
* **IOError**: if the file can’t be opened
* **KeyboardInterrupt**: when an unrequired key is pressed by the user
* **ValueError**: when built-in function receives a wrong argument
* **EOFError**: if End-Of-File is hit without reading any data
* **ImportError**: if it is unable to find the module


*ZeroDivisionError	Division by zero
*TypeError	Wrong data type
*IndexError	List index out of range
*KeyError	Dictionary key not found
*FileNotFoundError	File does not exist
*NameError	Variable not defined


**Difference between Syntax Error and Exceptions**
* ***Syntax Error:*** As the name suggests this error is caused by the wrong syntax in the code. It ***leads to the termination of the program.***

* ***Exceptions:*** Exceptions are raised when the program is syntactically correct, but the code resulted in an error. This error ***does not stop the execution of the program,*** however, it changes the normal flow of the program.

Why Use Exception Handling?

Prevents program crashes.

Handles errors gracefully.

Improves program reliability.

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
    print('result',result)

except ZeroDivisionError:
    print("Division by zero is not allowed.")

except ValueError:
    print("Please enter a valid integer.")
```

```python
try:
    num = int(input("Enter a number: "))

    if num < 0:
        raise ValueError("Negative numbers are not allowed.")
        # raise manually generates an exception

    result = 10 / num
    print("Result =", result)

except ZeroDivisionError:
    print("Division by zero is not allowed.")

except ValueError as e:
    print(e)
    
#except ValueError:
    # The exception is caught by the except ValueError block and the message is displayed.
   # print("Please enter a valid integer.")
```

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Result:", result)
    
    

#Using else

#The else block executes if no exception occurs.
```

```python
try:
    file = open("test.txt", "r")
    print(file.read())

except FileNotFoundError:
    print("File not found.")

finally:
    print("Program execution completed.")
#The finally block always executes whether an exception occurs or not.
```

<h3>3. Logical Error</h3>
A Logical Error occurs when a program runs without any syntax or runtime errors, but produces the wrong output because the logic of the program is incorrect

```python
# logical error
num = 7

if num % 2 == 1:
    print("Even Number")
else:
    print("Odd Number")
    
#The program executes without errors, but the condition is incorrect. It should be: num%2 ==0
    
#No SyntaxError
#No runtime exception
#Program executes normally
#Output is incorrect
#Harder to detect than syntax or runtime errors
```

<h3>assert statement and catching AssertionError. </h3>
assert condition, message

```python
# handling AssertionError
try:
    x = int(input('Enter a number between 15 and 40: '))
    assert x>=15 and x<=40
    print('The number entered: ', x)
except AssertionError:
    print('The condition is not fulfilled')
```

```python
# handling AssertionError
try:
    x = int(input('Enter a number between 15 and 40: '))
    assert x>=15 and x<=40, "Your input is not correct"
    print('The number entered: ', x)
except AssertionError as obj:
    print(obj)
```

<h3> our own exception and raise it when needed. </h3>

Minimum age requirement
Invalid account balance

Insufficient attendance
Negative quantity not allowed

```python
# User-defined exception
class NegativeNumberError(Exception):     # Create a custom exception class NegativeNumberError is a new exception type.
    pass                                   # It inherits from Python's built-in Exception class

try:
    num = int(input("Enter a positive number: "))

    if num < 0:
        raise NegativeNumberError("Negative numbers are not allowed.")  # raise manually generates the exception when the condition is met.

    print("You entered:", num)

except NegativeNumberError as e:   # Catches the custom exception and displays the error message.
    print("Error:", e)
    
finally: 
     print("We are not catching the exception.")
```
