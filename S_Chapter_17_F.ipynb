<h3>Working with Text Files Containing Strings.</h3>

```python
# creating a file with strings
# open the file for writing data
f = open('demofile.txt', 'w')

# enter strings from keyboard
print('Enter text (@ at end): ')

while str != '@':
    str = input() # accept string into str
    # write the string into file
    if(str != '@'):
        f.write(str+"\n")
        
# closing the file
f.close()
```

```python
# open the file for reading data
f = open('demofile.txt', 'r')

# read strings from the file
print('The file contents are:')

s = f.read()          # read entire file as a string
lines = s.splitlines() # split into lines

print(lines)

# closing the file
f.close()
```

```python
# appending and then reading strings and open the file for reading data
f = open('demofile.txt', 'a+')
print('Enter text to append(# at end): ')

while str != '#':
    str = input() # accept string into str
    # write the string into file
    if(str != '#'):
        f.write(str+"\n")
        
# put the file pointer to the beginning of file
f.seek(0,0)  # f.seek(offset,fromwhere)

# read strings from the file
print('The file contents are: ')

str = f.read()
print(str)

# closing the file
f.close()
```

```python
# counting number of lines, words and characters in a file
import os, sys

# open the file for reading data
fname = input('Enter filename: ')


if os.path.isfile(fname):
    f = open(fname, 'r')
else:
    print(fname+' does not exist')
    sys.exit()
    
    
# initialize the counters to 0
cl= cw= cc = 0

# read line by line from the file
for line in f:
    words = line.split()
    cl +=1
    cw += len(words)
    cc += len(line)
    
print('No. of lines: ', cl)
print('No. of words: ', cw)
print('No. of characters: ', cc)
# close the file
f.close()
```

**With statement**
* with statement in Python is used in exception handling to make the code cleaner and much more readable. 
* It simplifies the management of common resources like file streams. 
* Unlike the above implementations, ***there is no need to call file.close()*** when using with statement. 
* The with statement itself ensures proper acquisition and release of resources.
* **Syntax:** with open filename as file: with open("filename","openmode")


```python
with open('Images/python1.jpg', 'rb') as f1, open('copy.jpg', 'wb') as f2:
    f2.write(f1.read())

print("Image copied successfully")
```

## Pickle in Python


```python

#program to create an Emp class with employee details as instance variables

class Emp:
    def __init__(self, id, name, sal, mobile):
        self.id = id
        self.name = name
        self.sal = sal
        self.mobile = mobile
    def display(self):
        print("{:5d} {:20s} {:10.2f} {:10}".format(self.id, self.name,self.sal, self.mobile))

# pickle - store Emp class objects into emp.dat file

import pickle
# open emp.dat file as a binary file for writing
f = open('emp.dat', 'wb')
n = int(input('How many employees? '))
for i in range(n):
    id = int(input('Enter id: '))
    name = input('Enter name: ')
    sal = float(input('Enter salary: '))
    mobile = float(input('Enter mobile: '))
    
    # create Emp class object
    e = Emp(id, name, sal, mobile)
    
    # store the object e into the file f
    pickle.dump(e, f)
# close the file
f.close()
```

```python
# unpickle or object de-serialization
# import Emp, pickle

import pickle
# open the file to read objects


f = open('emp.dat', 'rb')
print('Employees details: ')

while True:
    try:
        # read object from file f/ unpickling is done using load()
        obj = pickle.load(f)
        # display the contents of employee obj
        obj.display()
    except EOFError:
        print('End of file reached...')
        break
        
# close the file
f.close()
```

<h3>Create phone book with names and phone numbers.
</h3>

```python
# create phonebook.dat file

# open the file in wb mode as binary file
with open("phonebook.dat", "wb") as f:
    
# wb mode: write bytesinto binary file use for writing purpose

    # write data into the file
    n = int(input('How many entries? '))

    for i in range(n):
        name = input('Enter name: ')
        phone = input('Enter phone no: ')

        # convert name and phone from strings to bytes
        name = name.encode()
        phone = phone.encode()

        # write the data into the file
        f.write(name + phone)
```

<h3>Random Accessing of binary files using mmap and performing various operations on a
binary file. </h3>

```python
# using mmap on a binary file, performing various operations
import mmap, sys

# display a menu
print('1 to display all the entries')
print('2 to display phone number')
print('3 to modify an entry')
print('4 exit')

ch = input('Your choice: ')

if ch == '4':
    sys.exit()

with open("phonebook.dat", "r+b") as f:
    # memory-map the file, size 0 means whole file
    mm = mmap.mmap(f.fileno(), 0)

    # display the entire file
    if ch == '1':
        print(mm.read().decode())

    # display phone number
    if ch == '2':
        name = input('Enter name: ')

        # find position of name
        n = mm.find(name.encode())

        # go to end of name
        n1 = n + len(name)

        # display the next 10 bytes
        ph = mm[n1: n1 + 10]
        print('Phone no: ', ph.decode())

    # modify phone number
    if ch == '3':
        name = input('Enter name: ')

        # find position of name
        n = mm.find(name.encode())

        # go to end of name
        n1 = n + len(name)

        # enter new phone number
        ph1 = input('Enter new phone number: ')

        # the old phone number is 10 bytes after n1
        mm[n1: n1 + 10] = ph1.encode()

    # close the mmap object
    mm.close()
```

## Zipping and Unzipping Files

```python
# zipping the contents of files
from zipfile import *

# create zip file
f = ZipFile('demo.zip', 'w', ZIP_DEFLATED)

# add some files
f.write('file1.txt')
f.write('file2.txt')
f.write('file3.txt')


# close the zip file
print('demo.zip file created...')
f.close()
```

```python
# to view contents of zipped files
from zipfile import *

# Open the zip file
z = ZipFile('demo.zip', 'r')

# extract all the file names which are in the zip file
z.extractall()
print("All files extracted successfully.")
```

```python
from zipfile import *

z = ZipFile('demo.zip', 'r')

for filename in z.namelist():
    print("Contents of", filename)
    print("-" * 20)

    data = z.read(filename).decode('utf-8')
    print(data)

    print()

z.close()
```
